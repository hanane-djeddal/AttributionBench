```python
import numpy as np
from statsmodels.stats.contingency_tables import mcnemar

def run_significance_test(model_ref, model_new, name="Model"):
    # Extract accuracy lists (1 for correct, 0 for incorrect)
    acc_ref = np.array([d['accuracy'] for d in model_ref])
    acc_new = np.array([d['accuracy'] for d in model_new])
    
    # Calculate the contingency table
    # n10: Ref correct, New wrong
    # n01: New correct, Ref wrong
    n10 = np.sum((acc_ref == 1) & (acc_new == 0))
    n01 = np.sum((acc_ref == 0) & (acc_new == 1))
    
    # McNemar's test only needs the discordant cells (n10 and n01)
    # We build the 2x2 table for the library
    n00 = np.sum((acc_ref == 0) & (acc_new == 0))
    n11 = np.sum((acc_ref == 1) & (acc_new == 1))
    table = [[n11, n10], 
             [n01, n00]]
    
    # exact=True is better for small datasets; exact=False uses Chi-squared
    result = mcnemar(table, exact=True)
    p_value = result.pvalue
    
    print(f"--- Comparing {name} against Baseline ---")
    print(f"P-value: {p_value:.10f}")
    
    # Adjusted thresholds for 2 comparisons (Bonferroni)
    # Original 0.05 becomes 0.025 | Original 0.01 becomes 0.005
    if p_value < 0.005:
        print("Result: Highly Significant (p < 0.01 level)")
    elif p_value < 0.025:
        print("Result: Significant (p < 0.05 level)")
    else:
        print("Result: Not statistically significant")
    print("\n")

# Usage:
# run_significance_test(results_m1, results_m2, name="Model 2")
# run_significance_test(results_m1, results_m3, name="Model 3")
```

```python
import numpy as np
from scipy.stats import pearsonr, spearmanr, kendalltau

def compute_correlations(results_list,gold_label="postprocess_label",precition="raw_output",align=False):
    """
    Computes correlation between 'ground_label' and 'label' 
    found in the list of dictionaries.
    """
    # 1. Extract and convert labels to numeric if they aren't already
    # This creates a mapping (e.g., 'Positive' -> 0, 'Negative' -> 1)
    # unique_labels = sorted(list(set([d['ground_label'] for d in results_list] + 
    #                                  [d['label'] for d in results_list])))
    #label_to_idx = {label: i for i, label in enumerate(unique_labels)}
    if align:
        y_pred=np.array([1 if r["auto_score"][0]>=0.6 else 0 for r in results_list])
        y_true = np.array([1 if r[gold_label] =="attributable" else 0  for r in results_list  ])
    elif gold_label=="attribution_label":
        y_true = np.array([1 if r[gold_label] =="attributable" else 0  for r in results_list  ])
        y_pred = [int(r[precition]) for r in results_list ]
        #print(y_pred,y_true)
    else:
        y_true = np.array([int(r[gold_label]) for r in results_list]) #[label_to_idx[d['postprocess_label']] for d in results_list])
        y_pred = [int(r[precition]) if r[precition] in ['0','1'] else None for r in results_list ]

    for it in range(len(y_pred)):
        if y_pred[it]== None:
            if '1' in results_list[it]:
                y_pred[it]=1
            elif '0' in results_list[it]:
                y_pred[it]=0
            else:
                y_pred[it]=0

    y_pred=np.array(y_pred)

    # 2. Compute Pearson
    p_corr, p_pvalue = pearsonr(y_true, y_pred)
    
    # 3. Compute Spearman
    s_corr, s_pvalue = spearmanr(y_true, y_pred)
    
    # 4. Compute Kendall
    k_corr, k_pvalue = kendalltau(y_true, y_pred)
    
    print(f"Correlation Analysis (N={len(y_true)}):")
    print(f"---")
    print(f"Pearson r  : {p_corr:.4f} (p={p_pvalue:.4f})")
    print(f"Spearman ρ : {s_corr:.4f} (p={s_pvalue:.4f})")
    print(f"Kendall τ  : {k_corr:.4f} (p={k_pvalue:.4f})")
    print(f"---")

# Usage:
```

```python
import numpy as np
from sklearn.metrics import f1_score
from sklearn.utils import resample
from tqdm import tqdm # Optional: for a progress bar

def bootstrap_significance(results_ref, results_new, n_iterations=1000, alpha=0.05):
    """
    results_ref: list of dicts for Model 1 (Baseline)
    results_new: list of dicts for Model 2 or 3
    """
    # 1. Extract ground truth and predictions
    y_true = np.array([d['ground_label'] for d in results_ref])
    y_pred_ref = np.array([d['label'] for d in results_ref])
    y_pred_new = np.array([d['label'] for d in results_new])
    
    # Calculate observed F1 difference
    obs_f1_ref = f1_score(y_true, y_pred_ref, average='macro')
    obs_f1_new = f1_score(y_true, y_pred_new, average='macro')
    observed_diff = obs_f1_new - obs_f1_ref
    
    print(f"Observed F1 Difference: {observed_diff:.4f} (New: {obs_f1_new:.4f}, Ref: {obs_f1_ref:.4f})")

    # 2. Bootstrapping loop
    diffs = []
    indices = np.arange(len(y_true))
    
    for i in range(n_iterations):
        # Sample indices with replacement
        boot_indices = resample(indices, replace=True, n_samples=len(y_true))
        
        # Calculate F1 for this bootstrap sample
        f1_ref = f1_score(y_true[boot_indices], y_pred_ref[boot_indices], average='macro')
        f1_new = f1_score(y_true[boot_indices], y_pred_new[boot_indices], average='macro')
        
        diffs.append(f1_new - f1_ref)
    
    diffs = np.array(diffs)
    
    # 3. Calculate P-value 
    # (How often is the difference <= 0, assuming the new model is better?)
    p_value = np.sum(diffs <= 0) / n_iterations
    
    # 4. Calculate 95% Confidence Interval for the difference
    ci_lower = np.percentile(diffs, 2.5)
    ci_upper = np.percentile(diffs, 97.5)
    
    print(f"P-value: {p_value:.4f}")
    print(f"95% CI for Difference: [{ci_lower:.4f}, {ci_upper:.4f}]")
    
    if p_value < 0.01:
        print("Status: Highly Significant (p < 0.01)")
    elif p_value < 0.05:
        print("Status: Significant (p < 0.05)")
    else:
        print("Status: Not Statistically Significant")
        
    return p_value, (ci_lower, ci_upper)

# Example Execution:
# p2, ci2 = bootstrap_significance(results_m1, results_m2, n_iterations=1000)
# p3, ci3 = bootstrap_significance(results_m1, results_m3, n_iterations=1000)
```
